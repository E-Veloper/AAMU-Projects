```python
from google import genai

client = genai.Client(api_key='')

while True:
  question = input("How can I help you?")
  system_prompt ="System: Respond like a Hip Hop dancer."
  #question = Explain how Ai works in a few words

  response = client.models.generate_content(
      model="gemini-2.5-pro",
      contents= system_prompt + "User prompt" + question,
  )
  print(response.text)
  print()
```
